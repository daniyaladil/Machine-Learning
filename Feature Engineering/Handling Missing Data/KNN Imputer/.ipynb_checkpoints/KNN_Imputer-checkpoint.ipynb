```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

from sklearn.impute import KNNImputer, SimpleImputer
```

```python
df=pd.read_csv("train.csv",usecols=["Survived","Age","Fare"])
df.sample(5)
```

```python
X= df.drop(columns=["Survived"])
Y=df["Survived"]
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X,Y, test_size=0.2, random_state=2)
```

# Using KNN Imputer

```python

knn=KNNImputer(weights='uniform',n_neighbors=4,add_indicator=True)
X_train_trans= knn.fit_transform(X_train)
X_test_trans=knn.transform(X_test)
```

```python
lr= LogisticRegression()
lr.fit(X_train_trans, Y_train)

y_pred= lr.predict(X_test_trans)

accuracy_score(Y_test, y_pred)
```

```python
np.mean(cross_val_score(lr, X_train_trans, Y_train, cv=100))
```

# Using Simple Imputer Class

```python

si=SimpleImputer()
X_train_trans= si.fit_transform(X_train)
X_test_trans=si.transform(X_test)
```

```python
lr= LogisticRegression()
lr.fit(X_train_trans, Y_train)

y_pred= lr.predict(X_test_trans)

accuracy_score(Y_test, y_pred)
```

```python
np.mean(cross_val_score(lr, X_train_trans, Y_train, cv=100))
```
